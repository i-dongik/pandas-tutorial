### `mysql+pymysql`

`mysql+pymysql`은 SQLAlchemy에서 MySQL 데이터베이스에 연결할 때 사용하는 연결 문자열. `mysql`은 **dialect**이고, `pymysql`은 **driver**.

#### 1. **Dialect (`mysql`)**

- **Dialect**는 SQLAlchemy가 어떤 종류의 데이터베이스와 연결할지 지정하는 부분.
- `mysql`은 SQLAlchemy가 **MySQL 데이터베이스**를 사용할 때 사용하는 **dialect**입니다.
- SQLAlchemy는 다양한 데이터베이스를 지원하며, 각 데이터베이스마다 고유의 dialect가 존재. 예를 들어:
  - `postgresql` : PostgreSQL 데이터베이스
  - `sqlite` : SQLite 데이터베이스
  - `mssql` : Microsoft SQL Server

#### 2. **Driver (`pymysql`)**

- **Driver**는 SQLAlchemy가 데이터베이스와 연결할 때 사용할 **Python 라이브러리**.
- `pymysql`은 MySQL 데이터베이스와 연결할 때 사용하는 드라이버로, **Pure Python으로 작성된 MySQL 클라이언트 라이브러리**.
- SQLAlchemy는 여러 드라이버를 지원하며, MySQL과 연결할 때 `pymysql` 외에도 `mysqlclient`나 `MySQLdb` 같은 드라이버를 사용할 수 있다.

#### 3. **`mysql+pymysql`의 의미**

- `mysql+pymysql`은 **SQLAlchemy가 MySQL 데이터베이스에 연결할 때 `pymysql` 드라이버를 사용하도록 지정**.
- 예를 들어, `mysql+pymysql://username:password@localhost:3306/mydatabase`와 같은 형식으로 연결 문자열을 작성하면, SQLAlchemy는 `pymysql`을 통해 MySQL 서버에 연결.

```python
import json
import pandas as pd
from sqlalchemy import create_engine

DB = "pandas_tutorial"

with open('/home/dongik/src/config.json', 'r') as config_file:
    config = json.load(config_file)['mysql'][DB]
    
username = config['username']
password = config['password']
host = config['host']
port = config['port']
database = 'pandas_tutorial'

connection_string = f'mysql+pymysql://{username}:{password}@{host}:{port}/{database}'

engine = create_engine(connection_string)

try:
    with engine.connect() as connection:
        print("Database connection successful!")
except Exception as e:
    print(f"Database connection failed: {e}")
```
